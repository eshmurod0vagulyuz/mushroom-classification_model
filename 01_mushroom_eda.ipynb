```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url="https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/mushrooms.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
df.describe(include='all').T
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
(df=='?').sum().sort_values(ascending=False)
```

```python
df = df.replace('?', np.nan)

df.isnull().sum().sort_values(ascending=False).head()
```

```python
df['stalk-root'] = df['stalk-root'].fillna('unknown')
```

```python
df.duplicated().sum()
```

### EDA

```python
df['class'].value_counts()
```

```python
sns.countplot(data=df, x='class')

plt.title("qoziqorinlar sinflari taqsimoti")
plt.xlabel("Qoziqorin sinfi")
plt.ylabel("Soni")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="odor", hue="class")

plt.title("Hid va qoziqorin sinfi")
plt.xlabel("Hid")
plt.ylabel("Soni")
plt.show()
```

```python
features = ['cap-shape','cap-color','bruises','gill-size',
            'ring-type','habitat']

fig, axes = plt.subplots(2, 3, figsize=(18,10))

for ax, col in zip(axes.flat, features):
    sns.countplot(data=df, x=col, hue='class', ax=ax)
    ax.set_title(f"{col}  va class")
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
